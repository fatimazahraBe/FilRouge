### data CPI inflation

```python
import requests
import pandas as pd

url = "https://api.worldbank.org/v2/country/all/indicator/FP.CPI.TOTL.ZG"
params = {
    "date": "2015:2025",
    "format": "json",
    "per_page": 20000
}

r = requests.get(url, params=params)
data = r.json()[1]

df = pd.DataFrame(data)
df = df[["countryiso3code", "country", "date", "value"]]

df.to_csv("worldbank_inflation_2015_2025.csv", index=False)
```
